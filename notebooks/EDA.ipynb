# 1. Data Understanding and Preprocessing

### 1. Exploratory Data Analysis (EDA):

```python
import pandas as pd

df = pd.read_csv('energy_data_set.csv')
```

```python
df.info()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt


# Preprocessing date for time-series analysis
df['date'] = pd.to_datetime(df['date'])
df['hour'] = df['date'].dt.hour
df['day_of_week'] = df['date'].dt.day_name()
df['month'] = df['date'].dt.month

# Line Chart - Energy Consumption Over Time
plt.figure(figsize=(15, 6))
plt.plot(df['date'], df['Appliances'], color='blue', linewidth=0.5)
plt.title('Appliances Energy Consumption Over Time')
plt.xlabel('Date')
plt.ylabel('Energy (Wh)')
plt.show()
```

```python
# Heatmap - Correlation Matrix
plt.figure(figsize=(12, 10))
corr = df.select_dtypes(include=['number']).corr()
sns.heatmap(corr, annot=False, cmap='coolwarm')
plt.title('Correlation Heatmap of Features')
plt.show()
```

```python
# Box Plot - Energy Consumption by Hour
plt.figure(figsize=(12, 6))
sns.boxplot(x='hour', y='Appliances', data=df)
plt.title('Energy Consumption Distribution by Hour of Day')
plt.show()
```

```python
# Box Plot - Energy Consumption by Day of Week
plt.figure(figsize=(12, 6))
order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
sns.boxplot(x='day_of_week', y='Appliances', data=df, order=order)
plt.title('Energy Consumption Distribution by Day of Week')
plt.show()
```

```python
# Histogram - Distribution of Appliances
plt.figure(figsize=(10, 6))
sns.histplot(df['Appliances'], bins=50, kde=True)
plt.title('Distribution of Appliances Energy Consumption')
plt.xlabel('Energy (Wh)')
plt.ylabel('Frequency')
plt.show()
```

```python
# Line Plot - Energy Consumption by Month
plt.figure(figsize=(12, 6))
sns.lineplot(x='month', y='Appliances', data=df.groupby('month')['Appliances'].mean().reset_index())
plt.title('Average Appliances Energy Consumption by Month')
plt.xlabel('Month')
plt.ylabel('Average Energy (Wh)')
plt.xticks(range(1, 13), ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
plt.grid(True)
plt.show()
```

```python
correlation_with_appliances = df.select_dtypes(include=['number']).corr()['Appliances'].sort_values(ascending=False)
display(correlation_with_appliances)
```

From the correlation values, we can observe that `RH_2`, `T2`, and `T_out` are some of the features with the strongest positive correlation with `Appliances` energy usage, while `RH_out` and `T6` show a strong negative correlation. These values indicate how much the features tend to change together (positive correlation) or in opposite directions (negative correlation).

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.scatterplot(x='RH_2', y='Appliances', data=df, alpha=0.6)
plt.title('Scatter Plot of Appliances vs. RH_2')
plt.xlabel('RH_2 (Relative Humidity in Living Room)')
plt.ylabel('Appliances Energy Consumption (Wh)')
plt.grid(True)
plt.show()
```
